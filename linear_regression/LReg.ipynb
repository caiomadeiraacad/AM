```python
import numpy as np
import math
import torch
```

```python
x = [1.0, 2.0]
y = [2.0, 4.0]
```

```python
w0 = 0
w1 = 0

alpha = 0.05

grad_w0 = 0.0
grad_w1 = 0.0

def model(x):
    return w0 + w1*x
```

```python
y_hats = []

for i in range(len(x)):
    y_hat = model(x[i])
    print(y_hat)

    error = y_hat - y[i]
    print(error)

    grad_w0 += 2*error 
    grad_w1 += 2*error*x[i]
    print(f"grad_w0={grad_w0}")
    print(f"grad_w1={grad_w1}")
```

```python
w0 = w0 - alpha * grad_w0
w1 = w1 - alpha * grad_w1
w0, w1
```
